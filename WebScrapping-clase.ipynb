```python
import requests
from bs4 import BeautifulSoup
import pandas as pd


url = "http://transparencia.uasnet.mx/2023/?seccion=2&subseccion=11"

data = {
    "unidad": "",
    "select_uo": "2700",
    "busqueda_por_uo": ""
}

response = requests.post(url, data=data)
response.encoding = "utf-8"

soup = BeautifulSoup(response.text, "html.parser")

# Extraer tabla de resultados
tabla = soup.find("table")
lista = []
if tabla:
    for fila in tabla.find_all("tr"):
        celdas = [td.get_text(strip=True) for td in fila.find_all(["td", "th"])]
        if celdas:
            print(celdas)
            #if celdas[2].__contains__("Patino"):
            if celdas[2].__contains__("M. ASIG.") or celdas[2].__contains__("PROF."):
                celdas.append(1)
            else:
                celdas.append(0)
            #print(type(celdas))
            lista.append(celdas)
           # df = pd.concat([df, celdas], ignore_index=True)
else:
    print("No se encontró tabla en la respuesta")


```
