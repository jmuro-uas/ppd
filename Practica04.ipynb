```python
import numpy as np

datos = np.array([1,12, 45, 7, 23, 67,67, 34, 89, 56, 11, 42])
```

## Medidas de tendencia central.

```python
print(np.mean(datos))       
print(np.median(datos))     
# Moda no está en NumPy, se usa scipy:
from scipy import stats
print(stats.mode(datos))
```

## Medidas de dispersión.

```python
print(np.std(datos))        # Desviación estándar
print(np.var(datos))        # Varianza
print(np.ptp(datos))        # Rango (max - min)
```

## Valores extremos y posición.

```python
print(np.min(datos))        # Mínimo
print(np.max(datos))        # Máximo
print(np.argmin(datos))     # Índice del mínimo
print(np.argmax(datos))     # Índice del máximo
print(np.percentile(datos, [25, 50, 75]))  # Cuartiles
```

## Operaciones sobre matrices.

```python
matriz = np.array([[10, 20, 30],
                   [40, 50, 60],
                   [70, 80, 90]])

print(np.mean(matriz))# Media
print(np.mean(matriz, axis=0))   # Media por columna
print(np.mean(matriz, axis=1))   # Media por fila
print(np.sum(matriz, axis=0))    # Suma por columna
```

## Correlación y Varianza

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 4, 5])

print(np.corrcoef(x, y))    # Matriz de correlación de Pearson
print(np.cov(x, y))         # Matriz de covarianza
```
