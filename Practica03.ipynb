## 1. Cargar los datos.

```python
import pandas as pd
import numpy as np

data = {
    "nombre":  ["Ana", "Luis", "luis", " María ", "Pedro", "Ana", "sofía"],
    "edad":    [25, 32, 32, "veintiocho", 150, 25, np.nan],
    "salario": [15000, 22000, 22000, 18500, 999999, 15000, np.nan],
    "ciudad":  ["Culiacán", "Mazatlán", "Mazatlán", "culiacán", np.nan, "Culiacán", "Los Mochis"],
    "fecha":   ["2024-01-15", "2024/03/20", "2024/03/20", "15-04-2024", "2024-05-10", "2024-01-15", "2024-06-01"],
}

df = pd.DataFrame(data)
print(df)
print("\nInfo general:")
print(df.info())
```

## 2. Duplicados

```python
df_clean = df.drop_duplicates()
```

## 4. Duplicados parciales: mismo nombre sin importar mayúsculas.

Normalizar nombre , Unificar mayúsculas/minúsculas — decidir una convención y aplicarla a todo:

.str.lower() → "ana" (útil para comparar)  
.str.upper() → "ANA"  
.str.title() → "Ana García" (la más legible para nombres)  
.str.capitalize() → "Ana garcía" (solo la primera letra del string)  
  
Colapsar espacios internos — "Luis  Alberto" (doble espacio) → "Luis Alberto" con .str.replace(r"\s+", " ", regex=True)

```python


df_clean["nombre_norm"] = df_clean["nombre"].str.strip().str.lower()
dupes_nombre = df_clean.duplicated(subset="nombre_norm", keep=False)
print("\nDuplicados por nombre (normalizado):")
print(df_clean[dupes_nombre])
```

## 5.Eliminar duplicados parciales

```python
df_clean = df_clean.drop_duplicates(subset="nombre_norm").drop(columns=["nombre_norm"])

print(f"\nFilas originales: {len(df)} → Después: {len(df_clean)}")

```

## 11.Cambio de tipos de datos  

Los datos importados llegan casi siempre como object. Convertir al tipo correcto reduce memoria, habilita operaciones y evita errores silenciosos.

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "nombre":    ["Ana", "Luis", "María", "Pedro", "Sofía"],
    "edad":      ["25", "32", "28", "41", "36"],       # str
    "salario":   ["15,000", "22,000", "18,500", "31,000", "27,500"],  # str con coma
    "activo":    ["Sí", "No", "Sí", "Sí", "No"],      # bool como texto
    "depto":     ["TI", "RRHH", "TI", "Ventas", "RRHH"]  # categoría
})

print(df.dtypes)  #dtype object
print(df.memory_usage(deep=True))
```

```python
df
```

## 12. Entero: quitar coma de miles y convertir

```python

df["edad"]    = df["edad"].astype("int8")     
df["salario"] = (df["salario"]
                  .str.replace(",", "")
                  .astype("int32"))
```

```python
df
```

## 13. Booleano: mapear texto → True/False

```python
# 
df["activo"] = df["activo"].map({"Sí": True, "No": False})

```

## 14. Categoría: columnas con pocos valores únicos

```python

df["depto"] = df["depto"].astype("category") #Solo pandas
```

## 15. Verificar y comparar uso de memoria

```python

print(df.dtypes)
print(df.memory_usage(deep=True))
```
